# KOMAROVA_CORE — реестр поддержки плагинов и интеграций

Этот ноутбук — параллельное техническое представление реестра, опубликованного на главном портале и странице облачной сети. Наличие компонента в списке не означает production-подключение.

## 1. Две поверхности одного проекта

- Главный портал: https://alexsandrkomarov1-wq.github.io/-ArtBukaBot-/ — пользовательская работа с поколениями, событиями, медиа, AI и маршрутами.
- Облачная сеть: https://alexsandrkomarov1-wq.github.io/-ArtBukaBot-/cloud-network/ — реестр копий, провайдеры, целостность, статусы и синхронизация.

Обе поверхности используют единые `PERSON-ID`, `EVENT-ID`, `PLACE-ID`, `MATERIAL-ID` и общий backend/API. Главная страница не пишет в облака напрямую.

## 2. Графические редакторы и дизайн

| Инструмент | Роль | Статус |
|---|---|---|
| Adobe Photoshop (UXP/API) | Ретушь, слои, восстановление фото | RESEARCH |
| Adobe Lightroom | Каталогизация и цветокоррекция | RESEARCH |
| GIMP, Krita | Бесплатная локальная обработка | DESIGNED |
| Canva, Figma | Альбомы, обложки, макеты и прототипы | RESEARCH |

Редактор работает с копией. Результат регистрируется как новый MATERIAL-ID со ссылкой `derived_from_material_id`; ORIGINAL остаётся неизменным.

## 3. Видео, проигрыватели и монтаж

| Инструмент | Роль | Статус |
|---|---|---|
| HTML5 Video, Video.js | Встроенное воспроизведение | DESIGNED / RESEARCH |
| VLC | Локальное воспроизведение редких форматов | DESIGNED |
| FFmpeg / FFprobe | Транскодирование, метаданные, кадры и звук | DESIGNED; worker не развёрнут |
| Shotcut, Kdenlive, DaVinci Resolve | Настольный монтаж и цветокоррекция | RESEARCH |

Плеер воспроизводит файл; редактор создаёт новую версию. Приватные файлы выдаются только через защищённый backend.

## 4. ИИ, разговорный режим, голос и звук

- AI Router — провайдер-независимое подключение моделей.
- Разговорный ассистент — текстовый и голосовой диалог по разрешённым данным.
- Whisper/STT — распознавание речи, транскрипты и таймкоды.
- Piper/TTS — озвучивание историй и подсказок.
- WebRTC — голосовая сессия; RNNoise/FFmpeg — очистка аудио.

Микрофон включается только по явному действию пользователя. Фоновая запись запрещена. Голосовые данные и секреты моделей не должны попадать в публичный браузер.

## 5. ИИ-летописец: видео и туристические прогулки

**По видео:** оригинал → проверка файла → извлечение аудио/кадров → речь и таймкоды → черновик сцен → связи с PERSON/EVENT/PLACE → проверка человеком → сохранение отдельного материала.

**По прогулке:** импорт GPX/KML/GeoJSON или разрешённых GPS-метаданных → точки и остановки → сопоставление фото/видео по времени → черновик маршрута и рассказа → ручная проверка географии → сохранение с PLACE-ID/EVENT-ID.

ИИ не устанавливает личность людей, не объявляет неподтверждённое место фактом и не сочиняет семейную историю. Все рассказы остаются `AI draft`, пока человек не одобрит их.

## 6. Карты, геокодирование и навигация

- OpenStreetMap — базовые геоданные с обязательной атрибуцией.
- MapLibre GL JS / Leaflet — карта в интерфейсе и облегчённый fallback.
- Nominatim — поиск адресов; использовать с учётом usage policy и rate limits.
- OSRM / GraphHopper / Valhalla — маршрутизация после проверки профиля, условий и API.
- Google Earth — KML/KMZ и внешняя визуализация, не источник истины для PLACE-ID.
- GPX/KML/GeoJSON — переносимые форматы треков.

Точные домашние координаты и частные места не публикуются автоматически.

## 9. Общие правила совместимости

1. Каталог KOMAROVA_CORE — источник истины для метаданных; облака хранят байты и реплики.
2. Оригинал неизменяем; производная версия получает новый MATERIAL-ID и ссылку на оригинал.
3. Состояние облачной копии `available` выставляется только после read-back и совпадения SHA-256 и размера.
4. Секреты и refresh tokens находятся только на сервере.
5. Сбой одного провайдера не блокирует остальные; отключение не удаляет объекты автоматически.
6. GitHub Pages — статический UI, не backend, не секрет-хранилище и не AI worker.
7. Все интеграции остаются DESIGNED/RESEARCH/NOT_CONNECTED до реального end-to-end теста.
